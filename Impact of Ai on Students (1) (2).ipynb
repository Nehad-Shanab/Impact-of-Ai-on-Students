# **GBA**

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X1=df.drop(columns=['Post_Semester_GPA'])
y1 = df['Post_Semester_GPA']

X1_train, X1_test, y1_train, y1_test = train_test_split(X1, y1, test_size=0.2, random_state=42)

scaler1 = StandardScaler()
X1_train = scaler1.fit_transform(X1_train)
X1_test = scaler1.transform(X1_test)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score
model1=RandomForestRegressor(n_estimators=100,max_depth=10,random_state=42)
model1.fit(X1_train,y1_train)

print(model1.score(X1_train,y1_train))
print(model1.score(X1_test,y1_test))

y1_pred=model1.predict(X1_test)
mae1=mean_absolute_error(y1_test,y1_pred)
r2_1=r2_score(y1_test,y1_pred)
print('MAE:',mae1)
print('R2:',r2_1)
```

# **Prediction**

```python
import gradio as gr
import pandas as pd


def predict_two_models(
    pre_gpa,
    weekly_genai,
    trad_study,
    tool_div,
    ai_dependency,
    anxiety_level,
):
  try:
    # 1. تجهيز المدخلات الأساسية
    base_dict = {
        'Pre_Semester_GPA': pre_gpa,
        'Weekly_GenAI_Hours': weekly_genai,
        'Traditional_Study_Hours': trad_study,
        'Tool_Diversity': tool_div,
        'Perceived_AI_Dependency': ai_dependency,
        'Anxiety_Level_During_Exams': anxiety_level,
    }

    # --- نموذج 1: التنبؤ بالمعدل (GPA) ---
    df1 = pd.DataFrame([base_dict])
    if 'X1' in globals():
      for col in X1.columns:
        if col not in df1.columns:
          df1[col] = 0
      df1 = df1[X1.columns]
    df1_scaled = scaler1.transform(df1)
    pred_gpa = model1.predict(df1_scaled)[0]

    # --- نموذج 2: التنبؤ بالاحتراق النفسي (Burnout) ---
    df2 = pd.DataFrame([base_dict])
    if 'X2' in globals():
      for col in X2.columns:
        if col not in df2.columns:
          df2[col] = 0
      df2 = df2[X2.columns]
    df2_scaled = scaler2.transform(df2)
    pred_burnout = model2.predict(df2_scaled)[0]

    if hasattr(model2, 'predict_proba'):
      prob2 = model2.predict_proba(df2_scaled)[0]
      if len(prob2) > 1:
        burnout_msg = (
            f'🚨 خطر احتراق مرتفع (High Risk) - نسبة الاحتمال:'
            f' {prob2[1]*100:.1f}%'
            if pred_burnout == 1
            else (
                f'✅ حالة طبيعية أو منخفضة (Normal/Low Risk) - نسبة الأمان:'
                f' {prob2[0]*100:.1f}%'
            )
        )
      else:
        burnout_msg = f'الحالة: {pred_burnout}'
    else:
      burnout_msg = f'الحالة: {pred_burnout}'

    return f'{pred_gpa:.2f} / 4.0', burnout_msg

  except Exception as e:
    err_msg = f'خطأ: {str(e)}'
    return err_msg, err_msg


# تصميم واجهة الاستخدام النظيفة والمبسطة
demo = gr.Interface(
    fn=predict_two_models,
    inputs=[
        gr.Slider(1.0, 4.0, value=3.2, step=0.05, label='المعدل السابق (Pre-GPA)'),
        gr.Number(value=8, label='ساعات استخدام الذكاء الاصطناعي أسبوعياً'),
        gr.Number(value=12, label='ساعات الدراسة التقليدية أسبوعياً'),
        gr.Slider(
            1, 5, value=3, step=1, label='تنوع الأدوات المستخدمة (1 إلى 5)'
        ),
        gr.Slider(1, 5, value=2, step=1, label='مستوى الاعتماد على الـ AI (1 إلى 5)'),
        gr.Slider(
            1, 5, value=3, step=1, label='مستوى القلق وقت الامتحانات (1 إلى 5)'
        ),
    ],
    outputs=[
        gr.Textbox(label='🎯 المعدل المتوقع للفصل القادم (GPA)'),
        gr.Textbox(label='⚠️ تقييم حالة الاحتراق النفسي (Burnout Risk)'),
    ],
    title='🎓 نظام التنبؤ الذكي بأداء الطلاب',
    description=(
        'أدخل بيانات الطالب أدناه، وسيقوم الذكاء الاصطناعي بتحليله وإعطائك'
        ' المؤشرات الأساسية فوراً!'
    ),
)

demo.launch(share=True)
```
